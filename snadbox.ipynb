```python
import random
```

```python
def custom_mutate(individual):
    total_vehicles = 2000
    index = random.randrange(4)
    max_increase = min(100, total_vehicles - individual[index])

    # So that the max decrease is 0
    max_decrease = min(100, individual[index])

    change = random.randint(-max_decrease, max_increase)
    individual[index] += change

    # After the change we need to adjust the other elements, so that the sum of individual is not greater than MAX_FLEET_SIZE
    while not sum(individual) == total_vehicles:
        other_index = random.choice([i for i in range(4) if i != index])

        if individual[other_index] - change >= 0 & individual[other_index] - change <= total_vehicles:
            individual[other_index] -= change

    return individual
```

```python
res = custom_mutate([0, 0, 58, 1942])

print(res)
print(sum(res))
```

```python
def custom_crossover(ind1, ind2):
    size = len(ind1)  # Nehmen an, dass beide Individuen gleich lang sind
    cxpoint1 = random.randint(0, size - 1)
    cxpoint2 = random.randint(0, size - 1)

    if cxpoint1 > cxpoint2:
        cxpoint1, cxpoint2 = cxpoint2, cxpoint1

    print(f'1: {cxpoint1} und 2: {cxpoint2}')

    # Perform the crossover between the two points
    temp1 = ind1[:cxpoint1] + ind2[cxpoint1:cxpoint2] + ind1[cxpoint2:]
    temp2 = ind2[:cxpoint1] + ind1[cxpoint1:cxpoint2] + ind2[cxpoint2:]

    # Adjust function to ensure no negative numbers and total is 2000
    def adjust_to_total(individual, target_sum):
        current_sum = sum(individual)
        diff = target_sum - current_sum
        adjustments = [0] * size
        while diff != 0:
            for i in range(size):
                if diff > 0 and individual[i] + adjustments[i] < 2000:
                    adjustments[i] += 1
                    diff -= 1
                elif diff < 0 and individual[i] + adjustments[i] > 0:
                    adjustments[i] -= 1
                    diff += 1
                if diff == 0:
                    break

        for i in range(size):
            individual[i] += adjustments[i]

        return individual

    # Apply the adjustment to ensure the sums are 2000
    adjust_to_total(temp1, 2000)
    adjust_to_total(temp2, 2000)

    # Replace original individuals
    ind1[:], ind2[:] = temp1, temp2
    return ind1, ind2
```

```python
res = custom_crossover([300, 500, 0, 1300], [500, 50, 1350, 100])

print(res)
print(sum(res[0]), sum(res[1]))
```
