## Scooter vs. Car Mode Choice

```python
# Sample Car and Scooter Trips
scooter_samples = trips[(trips["vehicleType"] == "scooter")].sample(100000)
car_samples = trips[(trips["vehicleType"] == "car")].sample(100000)
```

```python
# Concat Sampled Trips
trip_sampled = pd.concat([scooter_samples, car_samples])
```

```python
# https://www.statsmodels.org/stable/discretemod.html 
import statsmodels.api as sm

X = sm.add_constant(trip_sampled["distance"])
Y = trip_sampled["vehicleType"]
mnl_model = sm.MNLogit(Y, X).fit()

# spector_data = sm.datasets.spector.load_pandas()
```

```python
# Print Choice Model Summary
mnl_model.summary()
```

```python
# Predict Outcome based on Distance 
mnl_model.predict(X)
```

```python
# We generate test data from 0 meter to 10.000 meter and visualize the 
# probability densities

distance      = list(range(1,15000))
visualization = pd.DataFrame(distance, columns = ["distance"])
visualization = sm.add_constant(visualization)

# Predict Outcome
result = mnl_model.predict(visualization)

# Rename Columns 
result.columns = ["car", "scooter"]

# Copy Distance Columns
result["distance"] = visualization["distance"]
```

```python
import matplotlib.transforms as mtrans
from matplotlib.text import TextPath
from matplotlib.patches import PathPatch

# Curly Bracket 
def curly(x,y, scale, ax=None):
    if not ax: ax=plt.gca()
    tp = TextPath((0, 0), "}", size=1)
    trans = mtrans.Affine2D().scale(1, scale) + \
        mtrans.Affine2D().translate(x,y) + ax.transData
    pp = PathPatch(tp, lw=0, fc="k", transform=trans)
    ax.add_artist(pp)
```

# Plot NonStationary Demand 

```python
# Read Trips
demand = pd.read_csv('/Users/demircanm/Development/urban-mobility-simulation/data/demand/weekday_time_bucket_4h_h3_start_7.csv')
```

```python
from shapely.geometry import Polygon

# Convert H3 To Geometry
def add_geometry(row):
    points = h3.h3_to_geo_boundary(row['h3_start_7'], True)
    return Polygon(points)
```

```python
# Arrival Times 
df_1 = demand[(demand["is_holiday"]==False)&(demand["time_bucket_4h"]==2)].groupby("h3_start_7").agg({"o_poisson_est":"mean"}).reset_index()



#Apply function into our dataframe
df_1['geometry'] = (df_1.apply(add_geometry,axis=1))

gdf_1 = gpd.GeoDataFrame(df_1)
```

```python
# Arrival Times 
df_2 = demand[(demand["is_holiday"]==False)&(demand["time_bucket_4h"]==4)].groupby("h3_start_7").agg({"o_poisson_est":"mean"}).reset_index()


#Apply function into our dataframe
df_2['geometry'] = (df_2.apply(add_geometry,axis=1))

gdf_2 = gpd.GeoDataFrame(df_2)
```

```python
import geoplot as gplt
import geoplot.crs as gcrs

#import mapclassify
import contextily as ctx


proj = gcrs.WebMercator()
extent = (13.1027767, 52.3521867, 13.6021029, 52.653553)

f = plt.figure(figsize=(16,8))

ax1 = plt.subplot(121, projection=proj)
ax2 = plt.subplot(122, projection=proj)

ax1.set_title("a) Poisson Demand Process (00am-04am)", fontsize=16, fontname="Arial", x =0.53)
ax2.set_title("b) Poisson Demand Process (04pm-08pm)", fontsize=16, fontname="Arial", x = 0.55)

axis = [ax1, ax2]

# Get Max 
max_value = max(df_1["o_poisson_est"].max(), df_2["o_poisson_est"].max())

# kwargs for legend:
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.colorbar.html
gplt.choropleth(df=gdf_1,
                projection=proj,
               hue='o_poisson_est',
               figsize=(6,3),
               cmap='RdBu',
               legend = True,
               legend_kwargs = {"shrink":0.001, "ticks":[]},
               ax = ax1, 
               norm = matplotlib.colors.Normalize(vmin=0, vmax=max_value, clip=False)
)

gplt.choropleth(df=gdf_2,
                projection=proj,
               hue='o_poisson_est',
               figsize=(6,3),
               cmap='RdBu',
               legend = True,
               legend_kwargs = {"shrink":0.86},
               ax = ax2
)

gplt.webmap(gdf_1, extent = extent, projection=proj, ax=ax1, provider=ctx.sources.ST_TONER)

gplt.webmap(gdf_2, extent = extent, projection=proj, ax=ax2, provider=ctx.sources.ST_TONER)
    
test = ax1.legend()    

type(test)
# Adjust spacing between subplots


f.suptitle('Visualization of Non-Stationary Mobility Demand', fontsize=16, y=0.93)

f.subplots_adjust(left=None, bottom=None, right=None, top=None, wspace=-0.1, hspace=-0.9)
```
